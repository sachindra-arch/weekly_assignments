### Task 1: Sequence Generation (1 to 10)

### Task 4: Factorial Computation

This task requires implementing a script to find the product of all integers from 1 to $n$. For $n=5$, the output should be 120. We will use a `for` loop to calculate the factorial.

```python
# 4. Factorial Computation
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        product = 1
        for i in range(1, n + 1):
            product *= i
        return product

n_fact = 5 # Example value for n
print(f"\nFactorial of {n_fact} is: {factorial(n_fact)}")
```

### Task 5: The Fibonacci Sequence

For this task, we need to generate the first $n$ terms of the Fibonacci series ($0, 1, 1, 2, 3...$) using variable swapping. The Fibonacci sequence starts with 0 and 1, and each subsequent number is the sum of the two preceding ones.

```python
# 5. The Fibonacci Sequence
def fibonacci_sequence(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    else:
        fib_series = [0, 1]
        a, b = 0, 1
        for _ in range(2, n):
            a, b = b, a + b
            fib_series.append(b)
        return fib_series

n_fib = 10 # Example value for n
print(f"\nFibonacci sequence up to {n_fib} terms: {fibonacci_sequence(n_fib)}")
```

## Module 2: Number Theory & Logic

### Task 6: Prime Number Validation

This task involves writing logic to determine if a number $n$ is prime by checking for divisors between 2 and $n$. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
# 6. Prime Number Validation
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

num_prime = 17 # Example value for n
print(f"\nIs {num_prime} a prime number? {is_prime(num_prime)}")

num_not_prime = 15
print(f"Is {num_not_prime} a prime number? {is_prime(num_not_prime)}")
```

### Task 7: Integer Reversal

For this task, we will use a `while` loop and the modulo operator (`%`) to reverse the digits of any integer (e.g., $123 \to 321$). We'll extract digits one by one from the right.

```python
# 7. Integer Reversal
def reverse_integer(n):
    reversed_num = 0
    sign = 1
    if n < 0:
        sign = -1
        n = -n

    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n //= 10
    return sign * reversed_num

num_to_reverse = 12345 # Example value
print(f"\nReversed integer of {num_to_reverse} is: {reverse_integer(num_to_reverse)}")

num_to_reverse_neg = -678
print(f"Reversed integer of {num_to_reverse_neg} is: {reverse_integer(num_to_reverse_neg)}")
```

### Task 8: Numerical Palindromes

This task involves comparing an original number with its reversed version to see if they are identical. We can reuse the `reverse_integer` function from the previous task.

```python
# 8. Numerical Palindromes
def is_palindrome(n):
    # Reusing the reverse_integer function
    return n == reverse_integer(n)

palindrome_num = 121 # Example palindrome
print(f"\nIs {palindrome_num} a palindrome? {is_palindrome(palindrome_num)}")

non_palindrome_num = 123
print(f"Is {non_palindrome_num} a palindrome? {is_palindrome(non_palindrome_num)}")
```

### Task 9: Armstrong Numbers

This task requires checking if the sum of the cubes of a number's digits equals the number itself (e.g., $1^3 + 5^3 + 3^3 = 153$). We will extract digits and calculate their cubes.

```python
# 9. Armstrong Numbers
def is_armstrong_number(n):
    # Convert number to string to easily get digits and count them
    str_n = str(n)
    num_digits = len(str_n)
    sum_of_powers = 0
    temp_n = n

    while temp_n > 0:
        digit = temp_n % 10
        sum_of_powers += digit ** num_digits
        temp_n //= 10

    return n == sum_of_powers

armstrong_num = 153 # Example Armstrong number
print(f"\nIs {armstrong_num} an Armstrong number? {is_armstrong_number(armstrong_num)}")

non_armstrong_num = 123 # Example non-Armstrong number
print(f"Is {non_armstrong_num} an Armstrong number? {is_armstrong_number(non_armstrong_num)}")
```

### Task 10: Euclidean Algorithm (GCD)

Finally, for this task, we will use a `while` loop to find the Greatest Common Divisor (GCD) of two numbers through remainder division, implementing the Euclidean Algorithm. The GCD of two integers is the largest positive integer that divides both integers without a remainder.

```python
# 10. Euclidean Algorithm (GCD)
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

num1_gcd = 48
num2_gcd = 18
print(f"\nThe GCD of {num1_gcd} and {num2_gcd} is: {gcd(num1_gcd, num2_gcd)}")

num3_gcd = 101
num4_gcd = 103
print(f"The GCD of {num3_gcd} and {num4_gcd} is: {gcd(num3_gcd, num4_gcd)}")
```
