### 1. Recursive Fibonacci Sequence

````python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)

# Test the function
print(f"Fibonacci of 6: {fibonacci_recursive(6)}")
print(f"Fibonacci of 10: {fibonacci_recursive(10)}")
````

````python
def fibonacci_memoized(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    else:
        result = fibonacci_memoized(n-1, memo) + fibonacci_memoized(n-2, memo)
        memo[n] = result
        return result

# Test the memoized function
print(f"Memoized Fibonacci of 6: {fibonacci_memoized(6)}")
print(f"Memoized Fibonacci of 10: {fibonacci_memoized(10)}")
print(f"Memoized Fibonacci of 30: {fibonacci_memoized(30)}")

# Reset memo for another run if needed
# print(f"Memoized Fibonacci of 30 (new run): {fibonacci_memoized(30, {})}")
````

### 3. Product of a List using `functools.reduce()`

````python
from functools import reduce
import operator

# Define a list of numbers
numbers = [1, 2, 3, 4, 5]

# Use reduce with operator.mul to find the product
product = reduce(operator.mul, numbers)

print(f"The list of numbers is: {numbers}")
print(f"The product of the list using reduce is: {product}")

# Example with an empty list or single element (reduce has an optional initializer for this)
numbers_empty = []
numbers_single = [7]

# For an empty list without an initializer, reduce raises a TypeError.
# We can provide an initial value (e.g., 1 for product) to handle this.
product_empty = reduce(operator.mul, numbers_empty, 1)
product_single = reduce(operator.mul, numbers_single)

print(f"Product of empty list (with initializer 1): {product_empty}")
print(f"Product of single element list [7]: {product_single}")
````

### 4. Armstrong Number Check

````python
def is_armstrong_number(number):
    # Convert number to string to easily get its digits and length
    num_str = str(number)
    num_digits = len(num_str)

    sum_of_powers = 0
    for digit_char in num_str:
        digit = int(digit_char)
        sum_of_powers += digit ** num_digits

    return sum_of_powers == number

# Test the function
print(f"Is 9 an Armstrong number? {is_armstrong_number(9)}")
print(f"Is 10 an Armstrong number? {is_armstrong_number(10)}")
print(f"Is 153 an Armstrong number? {is_armstrong_number(153)}") # 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153
print(f"Is 370 an Armstrong number? {is_armstrong_number(370)}") # 3^3 + 7^3 + 0^3 = 27 + 343 + 0 = 370
print(f"Is 371 an Armstrong number? {is_armstrong_number(371)}") # 3^3 + 7^3 + 1^3 = 27 + 343 + 1 = 371
print(f"Is 1634 an Armstrong number? {is_armstrong_number(1634)}") # 1^4 + 6^4 + 3^4 + 4^4 = 1 + 1296 + 81 + 256 = 1634
print(f"Is 1234 an Armstrong number? {is_armstrong_number(1234)}")
````

### 5. Closure for Adding a Base Value

````python
def make_adder(base_value):
    # This is the outer function. base_value is its local variable.

    def adder(input_value):
        # This is the inner function, also known as the closure.
        # It 'remembers' or 'closes over' the base_value from its enclosing scope.
        return base_value + input_value

    return adder

# Create closures with different base values
add_five = make_adder(5)
add_ten = make_adder(10)

# Use the created closures
print(f"Using add_five(3): {add_five(3)}")   # Expected: 5 + 3 = 8
print(f"Using add_five(10): {add_five(10)}") # Expected: 5 + 10 = 15
print(f"Using add_ten(7): {add_ten(7)}")     # Expected: 10 + 7 = 17
print(f"Using add_ten(0): {add_ten(0)}")     # Expected: 10 + 0 = 10

# Demonstrate that 'base_value' is remembered for each closure
print(f"add_five remembers its base_value: {add_five.__closure__[0].cell_contents}")
print(f"add_ten remembers its base_value: {add_ten.__closure__[0].cell_contents}")
````

### 6. Explanation of the 'LEGB' Rule in Python Scope

**LEGB** stands for:

*   **L**ocal
*   **E**nclosing function locals
*   **G**lobal
*   **B**uilt-in

This rule dictates the order in which Python's interpreter looks for a name (variable, function, class, etc.) when it's referenced in code. It searches in the following order:

1.  **Local (L):** Python first checks the innermost scope, which is the current function or class method where the name is being referenced. This includes variables defined within the function.

2.  **Enclosing Function Locals (E):** If the name is not found in the local scope, Python then looks in the local scope of any enclosing (or outer) functions. This is particularly relevant for nested functions and closures, where an inner function can access variables from its outer function's scope (as seen in the `make_adder` example).

3.  **Global (G):** If the name is not found in any enclosing local scopes, Python moves to the global scope. This includes names defined at the top level of a module (a `.py` file), or explicitly declared as `global` within a function.

4.  **Built-in (B):** Finally, if the name is still not found, Python checks the built-in scope. This scope contains all the names that are pre-loaded into Python when the interpreter starts (e.g., `print`, `len`, `int`, `True`, `False`, `None`).

If the name is not found in any of these scopes, a `NameError` is raised.

#### Example of LEGB:

```python
x = 'global'

def outer_function():
    x = 'enclosing'

    def inner_function():
        x = 'local'
        print(x) # L: 'local'

    inner_function()
    print(x) # E: 'enclosing' (if inner_function didn't define x, it would be 'enclosing')

outer_function()
print(x) # G: 'global'

# Example of Built-in
# print() is always available
```

### 7. Decimal to Binary Conversion (without `bin()`)

````python
def decimal_to_binary(decimal_num):
    if decimal_num == 0:
        return "0"

    binary_string = ""
    temp_num = decimal_num

    while temp_num > 0:
        remainder = temp_num % 2
        binary_string = str(remainder) + binary_string
        temp_num //= 2

    return binary_string

# Test the function
print(f"Decimal 0 to Binary: {decimal_to_binary(0)}")
print(f"Decimal 1 to Binary: {decimal_to_binary(1)}")
print(f"Decimal 2 to Binary: {decimal_to_binary(2)}")
print(f"Decimal 10 to Binary: {decimal_to_binary(10)}") # Expected: 1010
print(f"Decimal 42 to Binary: {decimal_to_binary(42)}") # Expected: 101010
print(f"Decimal 255 to Binary: {decimal_to_binary(255)}") # Expected: 11111111
````

### 8. Generator for Infinite Even Numbers

````python
def infinite_even_numbers():
    num = 0
    while True:
        yield num
        num += 2

# Create a generator object
even_gen = infinite_even_numbers()

# Get the first few even numbers
print(f"First even number: {next(even_gen)}") # 0
print(f"Second even number: {next(even_gen)}") # 2
print(f"Third even number: {next(even_gen)}")  # 4

# Get the next 5 even numbers using a loop
print("\nNext 5 even numbers:")
for _ in range(5):
    print(next(even_gen))

# You can keep calling next() indefinitely to get more even numbers
````

### 9. Passing a Function as an Argument

````python
import math

def apply_operation(func, value):
    """Applies a given function to a value."""
    return func(value)

def square(x):
    return x * x

def cube(x):
    return x * x * x

# Passing 'square' function as an argument
result_square = apply_operation(square, 4)
print(f"Applying square to 4: {result_square}") # Expected: 16

# Passing 'cube' function as an argument
result_cube = apply_operation(cube, 3)
print(f"Applying cube to 3: {result_cube}")   # Expected: 27

# Passing a built-in function (e.g., math.sqrt)
result_sqrt = apply_operation(math.sqrt, 25)
print(f"Applying math.sqrt to 25: {result_sqrt}") # Expected: 5.0

# Passing a lambda function as an argument
result_lambda = apply_operation(lambda x: x + 10, 5)
print(f"Applying lambda x: x + 10 to 5: {result_lambda}") # Expected: 15
````

### 10. Password Validation Function

````python
import re

def validate_password(password):
    """
    Validates a password based on the following criteria:
    - Minimum 8 characters
    - At least one digit
    - At least one symbol (non-alphanumeric character)
    """
    if len(password) < 8:
        return False, "Password must be at least 8 characters long."

    if not re.search(r"\d", password):
        return False, "Password must contain at least one digit."

    # Check for at least one symbol (non-alphanumeric and non-whitespace character)
    if not re.search(r"[^a-zA-Z0-9\s]", password):
        return False, "Password must contain at least one symbol."

    return True, "Password is valid."

# Test the function
print(validate_password("Password123!")) # Expected: True
print(validate_password("password"))      # Expected: False (too short, no digit, no symbol)
print(validate_password("P@ssword"))    # Expected: False (no digit)
print(validate_password("Password123"))   # Expected: False (no symbol)
print(validate_password("short!1"))     # Expected: False (too short)
print(validate_password("ValidPassw0rd!")) # Expected: True
````
