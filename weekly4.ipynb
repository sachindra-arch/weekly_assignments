## Python Programming Resources

### 1. Join a list of strings into a single sentence
The `join()` method combines list elements into one string using the specified separator.

```python
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(sentence) # Python is awesome
```

### 2. Remove duplicates from a list using a set
A set stores only unique values, automatically removing duplicates.

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = list(set(numbers))
print(unique_numbers)
```

### 3. Check if a specific key exists in a dictionary
Use the `'in'` keyword to test whether a key is present.

```python
student = {"name": "Alice", "age": 20}
if "name" in student:
    print("Key exists")
```

### 4. Difference between `is` and `==`
`==` compares values.
`is` compares object identity (whether two variables refer to the same object).

```python
a = [1, 2]
b = [1, 2]
print(a == b) # True
print(a is b) # False
```

### 5. Simple calculator using `if-elif-else`

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")
if op == "+":
    print(num1 + num2)
elif op == "-":
    print(num1 - num2)
elif op == "*":
    print(num1 * num2)
elif op == "/":
    print(num1 / num2)
else:
    print("Invalid operator")
```

### 6. Find the largest number in a list without using `max()`

```python
numbers = [12, 45, 7, 89, 34]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print("Largest:", largest)
```

### 7. Use the `.strip()` method on a messy string
`strip()` removes leading and trailing whitespace.

```python
text = " Hello, World! "
clean_text = text.strip()
print(clean_text)
```

### 8. Convert a float to an integer
The decimal part is truncated (not rounded).

```python
num = 12.89
result = int(num)
print(result) # 12
```

### 9. Multiplication table of a given number

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

### 10. Use of the `'in'` keyword in Python
The `'in'` keyword checks membership.
It is commonly used with lists, strings, tuples, sets, and dictionaries.

```python
print(3 in [1, 2, 3, 4]) # True
print("a" in "apple") # True
print("name" in {"name": 1}) # True
```
