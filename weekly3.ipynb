### Task 1: Write a `while` loop that terminates when a user types "exit".

```python
user_input = ""
while user_input.lower() != "exit":
    user_input = input("Type something (or 'exit' to quit): ")
    print(f"You typed: {user_input}")
print("Loop terminated. Goodbye!")
```

### Task 2: Create a dictionary of 5 students and their marks; print only those who passed (>35).

```python
student_marks = {
    "Alice": 45,
    "Bob": 30,
    "Charlie": 55,
    "David": 25,
    "Eve": 40
}

print("Students who passed (marks > 35):")
for student, marks in student_marks.items():
    if marks > 35:
        print(f"{student}: {marks}")
```

### Task 3: Use a list comprehension to square all numbers in a list.

```python
numbers = [1, 2, 3, 4, 5]
squared_numbers = [x**2 for x in numbers]
print(f"Original numbers: {numbers}")
print(f"Squared numbers (using list comprehension): {squared_numbers}")
```

### Task 4: How do you merge two dictionaries in Python 3.9+?

```python
# Python 3.9+ introduced the merge operator (|) for dictionaries
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
dict3 = {'a': 5, 'e': 6} # Note: 'a' will be overwritten by dict3

merged_dict_union = dict1 | dict2
print(f"Using union operator (dict1 | dict2): {merged_dict_union}")

merged_dict_update = dict1 | dict3
print(f"Using union operator with overlapping key (dict1 | dict3): {merged_dict_update}")

# Alternatively, you can still use the unpacking operator (**)
merged_dict_unpacking = {**dict1, **dict2}
print(f"Using unpacking operator ({'{**dict1, **dict2}'}): {merged_dict_unpacking}")
```

### Task 5: Write a script to find the frequency of each character in a string.

```python
from collections import Counter

input_string = "hello world"
char_frequency = Counter(input_string)

print(f"Original string: '{input_string}'")
print(f"Character frequency: {char_frequency}")

# Or a manual way without Counter
char_freq_manual = {}
for char in input_string:
    char_freq_manual[char] = char_freq_manual.get(char, 0) + 1
print(f"Character frequency (manual): {char_freq_manual}")
```

### Task 6: Demonstrate the `enumerate()` function in a `for` loop.

```python
fruits = ["apple", "banana", "cherry", "date"]

print("Iterating through a list with enumerate():")
for index, fruit in enumerate(fruits):
    print(f"Index {index}: {fruit}")

print("\nIterating with a custom start index for enumerate():")
for index, fruit in enumerate(fruits, start=1):
    print(f"Item number {index}: {fruit}")
```

### Task 7: Write a program to find the intersection of two sets.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Using the intersection() method
intersection_set_method = set1.intersection(set2)
print(f"Set 1: {set1}")
print(f"Set 2: {set2}")
print(f"Intersection (using .intersection() method): {intersection_set_method}")

# Using the & operator
intersection_set_operator = set1 & set2
print(f"Intersection (using & operator): {intersection_set_operator}")
```

### Task 8: Use `zip()` to combine two lists into a dictionary.

```python
keys = ["name", "age", "city"]
values = ["Alice", 30, "New York"]

# Combine using zip() and dict()
combined_dict = dict(zip(keys, values))

print(f"Keys list: {keys}")
print(f"Values list: {values}")
print(f"Combined dictionary: {combined_dict}")

# Example with different lengths (zip stops at the shortest list)
keys_short = ["item1", "item2"]
values_long = ["valueA", "valueB", "valueC"]
combined_short_keys = dict(zip(keys_short, values_long))
print(f"\nCombined with shorter keys list: {combined_short_keys}")
```

### Task 9: Create a 3x3 matrix using nested lists and access the middle element.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("3x3 Matrix:")
for row in matrix:
    print(row)

# Accessing the middle element (row 1, column 1 for a 0-indexed matrix)
middle_element = matrix[1][1]
print(f"\nMiddle element of the matrix: {middle_element}")
```

### Task 10: Write a script to sort a dictionary by its values.

```python
my_dict = {'apple': 3, 'banana': 1, 'cherry': 2, 'date': 4}

print(f"Original dictionary: {my_dict}")

# Sort by values in ascending order
sorted_by_value_asc = sorted(my_dict.items(), key=lambda item: item[1])
print(f"Sorted by value (ascending): {sorted_by_value_asc}")

# Sort by values in descending order
sorted_by_value_desc = sorted(my_dict.items(), key=lambda item: item[1], reverse=True)
print(f"Sorted by value (descending): {sorted_by_value_desc}")

# Convert back to a dictionary (if desired, though it might lose ordering in older Python versions)
sorted_dict_asc = dict(sorted_by_value_asc)
print(f"Sorted dictionary (ascending by value): {sorted_dict_asc}")
```
