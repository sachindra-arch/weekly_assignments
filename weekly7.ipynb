## Object-Oriented Programming (OOP) Concepts in Python

### 1. Define a class `Car` with attributes and a method to display details.

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_details(self):
        return f"Make: {self.make}, Model: {self.model}, Year: {self.year}"

# Create an instance of Car
my_car = Car("Toyota", "Camry", 2020)
print(my_car.display_details())
```

### 2. Explain the purpose of the `__init__` method.

The `__init__` method is a special method in Python classes, often referred to as the **constructor**. Its primary purpose is to **initialize the attributes of an object** when a new instance of the class is created. When you create an object (e.g., `my_car = Car("Toyota", "Camry", 2020)`), Python automatically calls the `__init__` method for that object, passing the arguments provided to the method to set up the object's initial state.

### 3. Create a subclass `ElectricCar` that inherits from `Car`.

```python
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year) # Call the constructor of the parent class
        self.battery_size = battery_size

    def get_battery_info(self):
        return f"Battery size: {self.battery_size} kWh"

# Create an instance of ElectricCar
my_ev = ElectricCar("Tesla", "Model 3", 2023, 75)
print(my_ev.display_details())
print(my_ev.get_battery_info())
```

### 4. Demonstrate "Method Overriding" in inheritance.

Method overriding occurs when a subclass provides a specific implementation for a method that is already defined in its parent class. This allows the subclass to change the behavior of the inherited method.

```python
class ElectricCar(Car):
    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)
        self.battery_size = battery_size

    def display_details(self): # Overriding the display_details method from Car
        base_details = super().display_details()
        return f"{base_details}, Battery: {self.battery_size} kWh"

# Create an instance of ElectricCar
my_ev_override = ElectricCar("Nissan", "Leaf", 2022, 62)
print(my_ev_override.display_details()) # Calls the overridden method
```

### 5. Explain "Encapsulation" by creating private attributes using `__`.

Encapsulation is one of the fundamental principles of OOP. It refers to the bundling of data (attributes) and methods that operate on the data into a single unit (a class). It also restricts direct access to some of an object's components, which means that the internal representation of an object is hidden from the outside. In Python, you can achieve a form of encapsulation using **name mangling** by prefixing an attribute name with two leading underscores (e.g., `__private_attribute`). This makes the attribute not directly accessible from outside the class.

```python
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}. New balance: ${self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ${amount}. New balance: ${self.__balance}")
        else:
            print("Invalid withdrawal amount or insufficient funds.")

    def get_balance(self):
        return self.__balance

my_account = BankAccount(1000)
# print(my_account.__balance) # This would raise an AttributeError
print(f"Current balance: ${my_account.get_balance()}")
my_account.deposit(200)
my_account.withdraw(500)
print(f"Current balance: ${my_account.get_balance()}")
```

### 6. Use `@property` to create getter and setter methods.

The `@property` decorator provides a Pythonic way to use getters and setters without explicitly calling them as methods. It allows you to access methods like attributes, providing controlled access to an object's data.

```python
class Person:
    def __init__(self, name, age):
        self._name = name # Using single underscore, conventionally protected
        self._age = age

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        if isinstance(new_name, str) and len(new_name.strip()) > 0:
            self._name = new_name
        else:
            print("Invalid name.")

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, new_age):
        if isinstance(new_age, int) and 0 < new_age < 120:
            self._age = new_age
        else:
            print("Invalid age.")

p = Person("Alice", 30)
print(f"Name: {p.name}, Age: {p.age}")

p.name = "Bob" # Calls the setter
p.age = 25     # Calls the setter
print(f"New Name: {p.name}, New Age: {p.age}")

p.age = -5 # Invalid age, setter prevents update
print(f"Age after invalid attempt: {p.age}")
```

### 8. Create a class method using the `@classmethod` decorator.

A class method is a method that is bound to the class and not the instance of the class. It receives the class itself as the first argument, conventionally named `cls`. Class methods are often used as alternative constructors or to access/modify class-level attributes.

```python
class Pizza:
    radius = 5 # Class attribute

    def __init__(self, ingredients):
        self.ingredients = ingredients

    def get_ingredients(self):
        return self.ingredients

    @classmethod
    def margherita(cls):
        return cls(['mozzarella', 'tomatoes']) # cls refers to the Pizza class

    @classmethod
    def prosciutto(cls):
        return cls(['mozzarella', 'tomatoes', 'ham', 'mushrooms'])

print(f"Default pizza radius: {Pizza.radius}")

p1 = Pizza.margherita() # Calls the class method as an alternative constructor
p2 = Pizza.prosciutto()

print(f"Margherita ingredients: {p1.get_ingredients()}")
print(f"Prosciutto ingredients: {p2.get_ingredients()}")
```

### 9. Explain the difference between a class method and a static method.

**Class Method (`@classmethod`)**
*   **First Argument:** Takes `cls` (the class itself) as its first argument.
*   **Purpose:** Can access or modify class state (class variables). Often used for alternative constructors or factory methods that return an instance of the class.
*   **Use Case:** When you need to interact with the class itself, rather than a specific instance.

**Static Method (`@staticmethod`)**
*   **First Argument:** Does not take `self` (instance) or `cls` (class) as its first argument.
*   **Purpose:** Behaves like a regular function, but it belongs to the class's namespace. It doesn't have access to the class's state or instance's state.
*   **Use Case:** For utility functions that logically belong to the class but don't need any class-specific or instance-specific data. They are essentially functions defined inside a class.


```python
class MyClass:
    class_variable = "I am a class variable"

    def __init__(self, instance_variable):
        self.instance_variable = instance_variable

    @classmethod
    def class_method_example(cls):
        print(f"This is a class method. Accessing class variable: {cls.class_variable}")
        # Can also create an instance: return cls("new instance")

    @staticmethod
    def static_method_example(x, y):
        print(f"This is a static method. Performing calculation: {x} + {y} = {x + y}")
        # Cannot access class_variable or instance_variable directly

# Calling class method
MyClass.class_method_example()

# Calling static method
MyClass.static_method_example(10, 20)

# Instance calling methods
obj = MyClass("instance_data")
obj.class_method_example()
obj.static_method_example(5, 5)
```

### 10. Write a `try-except-finally` block to handle a `ZeroDivisionError`.

The `try-except-finally` block is used for error handling in Python.
*   The `try` block contains the code that might raise an exception.
*   The `except` block catches and handles specific exceptions.
*   The `finally` block contains code that will always be executed, regardless of whether an exception occurred or not (e.g., for cleanup operations).

```python
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        result = None # Or raise the exception again, or return a default value
    except TypeError:
        print("Error: Invalid input types. Please provide numbers.")
        result = None
    else:
        print("Division successful!")
    finally:
        print("--- Division attempt complete ---")
    return result

print("Attempt 1: 10 / 2")
print(f"Result: {divide_numbers(10, 2)}\n")

print("Attempt 2: 10 / 0")
print(f"Result: {divide_numbers(10, 0)}\n")

print("Attempt 3: 'a' / 2")
print(f"Result: {divide_numbers('a', 2)}\n")
```
